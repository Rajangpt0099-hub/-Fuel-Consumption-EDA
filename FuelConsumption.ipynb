# Outlier Analysis

```python
Q1 = df["CO2EMISSIONS"].quantile(0.25)
Q3 = df["CO2EMISSIONS"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

- The IQR method is used to identify potential CO2 emission outliers.

# Number of Outliers

```python
outliers = df[
    (df["CO2EMISSIONS"] < lower_bound) |
    (df["CO2EMISSIONS"] > upper_bound)
]

print("Number of outliers:", len(outliers))
```

- The analysis identifies vehicles with unusually high or low CO2 emissions.
